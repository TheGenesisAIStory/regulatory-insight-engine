Notebook zero-touch: monta Drive, armonizza la root operativa, verifica A100, esegue training, eval, report, release ZIP e demo Gradio pubblica solo se il verdetto reale e positivo.

```python
# Fiorell.IA Gold Release - run all
from pathlib import Path
import os
import sys
import urllib.request

RAW_BASE = "https://github.com/TheGenesisAIStory/regulatory-insight-engine/raw/main"
DRIVE_ROOT = Path("/content/drive/MyDrive")
REPO_ROOT = DRIVE_ROOT / "regulatory-insight-engine"

try:
    from google.colab import drive
except Exception as exc:
    raise RuntimeError("Apri questo notebook in Google Colab con runtime GPU A100.") from exc

drive.mount("/content/drive", force_remount=False)
REPO_ROOT.mkdir(parents=True, exist_ok=True)
script_path = REPO_ROOT / "fiorellia_gold_zero_touch.py"
script_url = f"{RAW_BASE}/fiorellia_gold_zero_touch.py"
script_path.write_text(urllib.request.urlopen(script_url).read().decode("utf-8"), encoding="utf-8")
os.chdir(REPO_ROOT)
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
os.environ["FIORELLIA_NOTEBOOK_NO_EXIT"] = "1"
namespace = {"__name__": "fiorellia_gold_zero_touch", "__file__": str(script_path)}
exec(compile(script_path.read_text(encoding="utf-8"), str(script_path), "exec"), namespace)
return_code = namespace["main"]()
print(f"Fiorell.IA Gold Release return_code={return_code}")
```
